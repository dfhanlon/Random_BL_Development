```python
import numpy as np
from larch.xray import xray_lines

def find_element_at_bin(bin_num, n_bins=4096, emax_eV=40960, threshold_eV=100):
    eV_per_bin = emax_eV / n_bins
    target_eV = bin_num * eV_per_bin
    print(f"Bin {bin_num} → {target_eV:.0f} eV  ({target_eV/1000:.3f} keV)\n")
    matches = []
    for elem in ["Ca","Ti","V","Cr","Mn","Fe","Co","Ni","Cu","Zn",
                 "As","Se","Br","Sr","Mo","Ag","Cd","Sn","Pb","Au"]:
        for line, data in xray_lines(elem).items():
            if abs(data.energy - target_eV) < threshold_eV:
                matches.append((abs(data.energy - target_eV), elem, line, data.energy))
    for diff, elem, line, energy in sorted(matches):
        print(f"  {elem} {line:10s}  {energy:.0f} eV  (Δ {diff:.0f} eV)")

find_element_at_bin(865)   # example: bin 804 → ~8040 eV → Cu Ka1
```
